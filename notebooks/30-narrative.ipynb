# The care gap in the Kingdom

"You might want to grab a mask for while you wait," suggested the woman at the reception pointing me to the waiting room. "There's a nasty new flu variant going around, and this is the place to get it."

By the time I checked in at the ExpressCare at the beginning of March, my immune system had already mastered the entire alphabet of flu, covid, and other variants that I came across every other week since Thanksgiving. For better or worse, I was finally clear of those seasonal illnesses and moving on to new maladies.

My visit was pleasant enough, and produced a prescription that seemed promising. But as much as I was relieved to get medication, I was bracing for the question that has gotten quite frustrating in recent years. "What pharmacy should we send this to?"

I live in Hardwick, Vermont. A small town of 3,000 in the state's rural Northeast Kingdom of Vermont. All my life the town had a pharmacy. When I was a kid it was called Brooks, at some point it changed to a Rite Aid, which eventually changed again to a Walgreens, before finally, [much beleagured](https://hardwickgazette.org/2025/11/18/walgreens-to-pay-state-again/), closing its doors in 2024. 

This left a critical gap in care in our community. Our nearest pharmacy is now in Morrisville, a 20 minute drive from the village of Hardwick. This has left thousands of people in Hardwick and beyond in a [pharmacy desert](https://www.japha.org/article/S1544-3191(24)00004-9/abstract).

## Mapping the desert

Hardwick is not the only underserved community. In fact, we can build a map to highlight all the underserved communities across the region.

Using [a tool I built](https://themarkup.org/still-loading/2023/03/02/introducing-the-united-states-place-sampler-usps) a couple years ago, I pulled addresses from across all of Northeastern Vermont (Essex, Caledonia, Orleans, Lamoille, and Washington counties). I then found all the active pharmacies anywhere in Northern Vermont from Google Maps. Finally, I estimated the driving time from every address to every pharmacy using [OSRM](https://project-osrm.org/) to find the minimum travel time to a pharmacy across the entire region.

Drawing these travel times on a map, we can see how important this pharmacy was to the area.

## The weight of the problem

There's a [quip](https://xkcd.com/1138/) that most attempts to show geographic trend maps are really just population maps. Ironically, the maps above don't show population, but really ought to.

As the maps show, the worst served area is certainly Essex County. A naive analysis would recommend putting a pharmacy in downtown Granby. And perhaps the 81 residents of Granby would welcome it--or, perhaps not. Either way, this would still leave tens of thousands of other people poorly served.

Drawing the population density gives a slightly better view of the problem. This map shows population density, with regions over 20 minutes from a pharmacy highlighted.

## The case for Hardwick

The Northeastern Vermont Regional Hospital was [recently awarded](https://www.welch.senate.gov/welch-joins-nek-health-care-providers-to-celebrate-1-4-million-in-new-federal-funding-to-help-build-pharmacies-in-caledonia-county/) a grant to address the care gap in Caledonia County, and in particular to bring a pharmacy back to Hardwick.

Selfishly, I welcome the news. But as we've seen, Hardwick is not the only community with a care gap. So I wonder, where is the optimal place to put a new pharmacy?

I reran the drive-time simulation with a new pharmacy in every town in the region. Since in Vermont a "town" is a sort of [umbrella administrative division](https://www.reddit.com/r/vermont/comments/16juhg9/comment/k0s5joy/) and not meaningful as a specific geographic point, I actually took [a list](https://github.com/jnu/farma/blob/main/data/vt_village_centers.csv) of 237 villages and other named places that could conceivably host a new pharmacy. For example, in Hardwick, we have the obvious location of the Hardwick village, the next-most-plausible one of East Hardwick village, and also some low-probability sites like Mackville and Hardwick Center. 

To account for population, I used a metric of human-minutes saved. In other words, multiplying the drive time to the new pharmacy by the number of residents who would use it, and then finding the aggregate time saved over their current travel time. Of course this isn't perfect for many reasons, since there are other variables we don't know that affect who is going to a pharmacy, how often, and which is most convenient. But this metric at least tries to estimate where a new pharmacy would do the most good for the most people.

Sure enough, the Hardwick village stands out as the most practical place to (re-)establish a pharmacy

## Where else is at risk?

Hardwick is not the only community in need. We can see there are still gaps in care that the Hardwick pharmacy won't fix. Small towns in even more rural area will still be 20 minutes from a pharmacy, like in Crafstbury, parts of Danville and Peacham. Island Pond could certainly use one, there are no pharmacies at all in Essex County.

In addition, using the same simulation method but with subtracting pharmacies one by one, we can see that the Kinney Drugs in Barton is carrying the most load. If that pharmacy closed, it would leave a care desert as big as the one that currently exists in Hardwick.
